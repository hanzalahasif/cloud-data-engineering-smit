```python
# Q41. Student Performance System

students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

averages = {
    students[0]["name"]: sum(students[0]["marks"]) / len(students[0]["marks"]),
    students[1]["name"]: sum(students[1]["marks"]) / len(students[1]["marks"]),
    students[2]["name"]: sum(students[2]["marks"]) / len(students[2]["marks"])
}
top_student = max(averages, key=averages.get)
names = tuple(student["name"] for student in students)
unique_marks = set(students[0]["marks"]) | set(students[1]["marks"]) | set(students[2]["marks"])

print("Averages:", averages)
print("Top student:", top_student)
print("Names:", names)
print("Unique marks:", unique_marks)
```

```python
# Q42. Employee Skills

employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

common_skills = employees["Ali"] & employees["Sara"] & employees["Ahmed"]
ali_only = employees["Ali"] - employees["Sara"] - employees["Ahmed"]
all_team_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]
python_sql = {
    name for name, skills in employees.items()
    if {"Python", "SQL"}.issubset(skills)
}

print("Common skills:", common_skills)
print("Ali-only skills:", ali_only)
print("All team skills:", all_team_skills)
print("Python + SQL:", python_sql)
```

```python
# Q43. Sales Data

sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

revenues = {
    sales[0][0]: sales[0][1] * sales[0][2],
    sales[1][0]: sales[1][1] * sales[1][2],
    sales[2][0]: sales[2][1] * sales[2][2]
}
top_revenue_product = max(revenues, key=revenues.get)
prices = {sale[1] for sale in sales}

print("Revenue:", revenues)
print("Top-revenue product:", top_revenue_product)
print("Price set:", prices)
```

```python
# Q44. Customer Orders

customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

spending = {
    customer_id: sum(data["orders"])
    for customer_id, data in customers.items()
}
top_customer = max(spending, key=spending.get)
totals = tuple(spending[customer_id] for customer_id in customers)

print("Spending:", spending)
print("Top customer:", top_customer)
print("Totals:", totals)
```

```python
# Q45. Course Enrollment

courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

all_students = courses["Python"] | courses["SQL"] | courses["AWS"]
all_three = courses["Python"] & courses["SQL"] & courses["AWS"]
exactly_two = (
    (courses["Python"] & courses["SQL"]) |
    (courses["Python"] & courses["AWS"]) |
    (courses["SQL"] & courses["AWS"])
) - all_three
exactly_one = all_students - (
    courses["Python"] & courses["SQL"] |
    courses["Python"] & courses["AWS"] |
    courses["SQL"] & courses["AWS"]
)

print("Exactly two:", exactly_two)
print("All three:", all_three)
print("Exactly one:", exactly_one)
print("Unique students:", len(all_students))
```

```python
# Q46. E-Commerce Transactions

transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

transaction_totals = {
    transactions[0]["id"]: sum(item[1] * item[2] for item in transactions[0]["items"]),
    transactions[1]["id"]: sum(item[1] * item[2] for item in transactions[1]["items"]),
    transactions[2]["id"]: sum(item[1] * item[2] for item in transactions[2]["items"])
}
total_revenue = sum(transaction_totals.values())
ali_spending = transaction_totals[101] + transaction_totals[103]
largest_transaction = max(transaction_totals, key=transaction_totals.get)
unique_products = (
    {item[0] for item in transactions[0]["items"]} |
    {item[0] for item in transactions[1]["items"]} |
    {item[0] for item in transactions[2]["items"]}
)

print("Total revenue:", total_revenue)
print("Ali spending:", ali_spending)
print("Largest transaction:", largest_transaction)
print("Unique products:", unique_products)
```

```python
# Q47. Pipeline Monitoring

pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

pipeline_totals = {
    name: sum(data["records"])
    for name, data in pipelines.items()
}
total_processed = sum(pipeline_totals.values())
failed_pipelines = {name for name, data in pipelines.items() if data["status"] == "failed"}
largest_pipeline = max(pipeline_totals, key=pipeline_totals.get)
averages = {
    name: sum(data["records"]) / len(data["records"])
    for name, data in pipelines.items()
}

print("Total processed:", total_processed)
print("Failed pipelines:", failed_pipelines)
print("Largest pipeline total:", largest_pipeline)
print("Average records per run:", averages)
```

```python
# Q48. Banking Transactions

accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

final_balances = {
    account_id: sum(data["transactions"])
    for account_id, data in accounts.items()
}
highest_balance = max(final_balances, key=final_balances.get)
total_withdrawals = -sum(
    transaction
    for data in accounts.values()
    for transaction in data["transactions"]
    if transaction < 0
)
largest_deposit = max(
    transaction
    for data in accounts.values()
    for transaction in data["transactions"]
    if transaction > 0
)

print("Final balances:", final_balances)
print("Highest balance account:", highest_balance)
print("Total withdrawals:", total_withdrawals)
print("Largest single deposit:", largest_deposit)
```

```python
# Q49. Product Recommendations

users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

purchased_by_at_least_two = (
    (users["Ali"]["purchased"] & users["Sara"]["purchased"]) |
    (users["Ali"]["purchased"] & users["Ahmed"]["purchased"]) |
    (users["Sara"]["purchased"] & users["Ahmed"]["purchased"])
)
ali_wishlist_minus_purchases = users["Ali"]["wishlist"] - users["Ali"]["purchased"]
shared_wishlist = (
    users["Ali"]["wishlist"] & users["Sara"]["wishlist"]
) | (
    users["Ali"]["wishlist"] & users["Ahmed"]["wishlist"]
) | (
    users["Sara"]["wishlist"] & users["Ahmed"]["wishlist"]
)
all_products = (
    users["Ali"]["purchased"] | users["Ali"]["wishlist"] |
    users["Sara"]["purchased"] | users["Sara"]["wishlist"] |
    users["Ahmed"]["purchased"] | users["Ahmed"]["wishlist"]
)

print("Bought by at least two:", purchased_by_at_least_two)
print("Ali wishlist minus purchases:", ali_wishlist_minus_purchases)
print("Shared wishlist products:", shared_wishlist)
print("All products:", all_products)
```

```python
# Q50. Final Company Analysis

company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

engineering = company["departments"]["Engineering"]
analytics = company["departments"]["Analytics"]

payroll = {
    "Engineering": sum(employee[1] for employee in engineering["employees"]),
    "Analytics": sum(employee[1] for employee in analytics["employees"])
}
company_payroll = sum(payroll.values())

all_employees = engineering["employees"] + analytics["employees"]
highest_paid = max(all_employees, key=lambda employee: employee[1])

shared_skills = (
    engineering["employees"][0][2] & engineering["employees"][1][2] &
    analytics["employees"][0][2] & analytics["employees"][1][2]
)
engineering_skills = engineering["employees"][0][2] | engineering["employees"][1][2]
analytics_skills = analytics["employees"][0][2] | analytics["employees"][1][2]
engineering_only = engineering_skills - analytics_skills
remaining_budget = {
    "Engineering": engineering["budget"] - payroll["Engineering"],
    "Analytics": analytics["budget"] - payroll["Analytics"]
}

average_salary = {
    "Engineering": payroll["Engineering"] / len(engineering["employees"]),
    "Analytics": payroll["Analytics"] / len(analytics["employees"])
}
higher_avg_department = max(average_salary, key=average_salary.get)

name_to_salary = {employee[0]: employee[1] for employee in all_employees}
all_unique_skills = engineering_skills | analytics_skills
combined_budget_sufficient = (
    engineering["budget"] + analytics["budget"] >= company_payroll
)

print("1. Payroll per department:", payroll)
print("2. Company payroll:", company_payroll)
print("3. Highest-paid employee:", highest_paid)
print("4. Skills shared by all:", shared_skills)
print("5. Engineering-only skills:", engineering_only)
print("6. Remaining budget:", remaining_budget)
print("7. Higher average salary department:", higher_avg_department)
print("8. Name-to-salary:", name_to_salary)
print("9. All unique skills:", all_unique_skills)
print("10. Combined budget sufficient:", combined_budget_sufficient)
```
