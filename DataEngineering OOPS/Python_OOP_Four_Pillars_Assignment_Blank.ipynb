Extension: Explain why double underscores do not provide true security.

```python
# Write your solution here
class BankAccount:
    def __init__(self,name,balance,pin):
        self.name = name
        self.__balance = balance
        self.__pin = pin
    def deposit(self,amount):
        if amount > 0:
            self.__balance  = self.__balance+ amount
            return self.__balance
        else:
            return "rejected"    
    def withdraw(self,amount, pin):
        if amount <= self.__balance and pin == self.__pin:
            return f" {amount} withdraw successful now balance is {self.__balance - amount} "
        else:
            return "get lost"
    def get_balance(self,pin):
        if self.__pin == pin:
            return self.__balance
        else:
            return "invalid pin"  

```

## Q7. Shape Calculator

Use ABC and @abstractmethod to define an abstract Shape class with area() and perimeter(). Implement both methods in Circle and Rectangle.

Accept dimensions through constructors and use pi = 3.14159. Test a circle with radius 5 and a rectangle with width 4 and height 6.

Attempt to instantiate Shape directly and explain the outcome.

```python
# Write your solution here
from abc import ABC , abstractmethod
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
    @abstractmethod
    def perimeter(self):
        pass
class Circle(Shape):
    pi = 3.14159
    def area(self,radius):
     
        return   self.pi * (radius * radius  )
    def perimeter(self,radius):
        
        return 2 * self.pi * radius
class Rectangle(Shape):
    def area(self, length, width):
        return length * width

    def perimeter(self, length, width):
        return 2 * (length + width)
```

```python
cicle=Circle()
cicle.area(5)
cicle.perimeter(5)
```
