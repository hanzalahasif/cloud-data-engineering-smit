# Python Practice — Tuples (Q11–Q20)

**Rules:** No `for`/`while` loops, comprehensions, recursion, `map`/`filter`/`reduce`, or hardcoded answers.

```python
# Q11. Tuple Indexing

colors = ("red", "green", "blue", "yellow", "purple")
second = colors[1]
last = colors[-1]
middle_three = colors[1:4]

print("Second:", second)
print("Last:", last)
print("Middle three:", middle_three)
```

```python
# Q12. Tuple Unpacking

employee = ("Ali", 28, "Data Engineer", 150000)

name, age, job, salary = employee

print("Name:", name)
print("Salary:", salary)
```

```python
# Q13. Tuple Concatenation

tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

combined = tuple1 + tuple2
reversed_tuple = combined[::-1]

print("Combined:", combined)
print("Reversed:", reversed_tuple)
```

```python
# Q14. Count and Locate

numbers = (10, 20, 30, 20, 40, 20, 50)

count_20 = numbers.count(20)
first_index = numbers.index(20)

print("Count of 20:", count_20)
print("First index of 20:", first_index)
```

```python
# Q15. Extended Tuple Unpacking

data = (100, 200, 300, 400, 500, 600)

first, second, *remaining, last = data
remaining_list = list(remaining)

print("First:", first)
print("Second:", second)
print("Last:", last)
print("Remaining:", remaining_list)
```

```python
# Q16. Modify an Immutable Tuple

numbers = (10, 20, 30, 40, 50)

new_numbers = numbers[:2] + (300,) + numbers[3:]

print("New tuple:", new_numbers)
```

```python
# Q17. Nested Tuple Access

students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

sara_second_score = students[1][1][1]
ali_average = sum(students[0][1]) / len(students[0][1])
ahmed_highest = max(students[2][1])

print("Sara second score:", sara_second_score)
print("Ali average:", ali_average)
print("Ahmed highest score:", ahmed_highest)
```

```python
# Q18. Tuple Sorting

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sorted_employees = tuple(sorted(employees, key=lambda employee: employee[1], reverse=True))

print("Sorted by salary descending:", sorted_employees)
```

```python
# Q19. Tuple Matrix

matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])
diagonal_product = diagonal[0] * diagonal[1] * diagonal[2]
row_sums = (sum(matrix[0]), sum(matrix[1]), sum(matrix[2]))

print("Diagonal:", diagonal)
print("Diagonal product:", diagonal_product)
print("Row sums:", row_sums)
```

```python
# Q20. Inventory Tuple Challenge

records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

laptop_value = records[0][1] * records[0][2]
mouse_value = records[1][1] * records[1][2]
keyboard_value = records[2][1] * records[2][2]

inventory_value = laptop_value + mouse_value + keyboard_value
most_expensive_product = max(records, key=lambda record: record[1])[0]
reversed_names = (records[2][0], records[1][0], records[0][0])

print("Inventory value:", inventory_value)
print("Most expensive product:", most_expensive_product)
print("Reversed names:", reversed_names)
```
