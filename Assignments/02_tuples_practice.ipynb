Extract Sara second score, Ali average, and Ahmed highest score.

Q18. Tuple Sorting

```python
employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

Sort by salary descending without loops, comprehensions, or lambda.


````python

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sort_emp=sorted(employees,key=lambda x:x[1],reverse=True)
sort_emp
````
