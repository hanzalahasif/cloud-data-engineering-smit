```python
# Q39
# Calculate each employee's average marks, find the top performer, and find the employee with the highest SQL mark.
employees = {
    "Ali": {"Python": 90, "SQL": 85, "AWS": 88},
    "Sara": {"Python": 95, "SQL": 92, "AWS": 90},
    "Ahmed": {"Python": 80, "SQL": 87, "AWS": 84}
}
ali_avrg=sum(employees["Ali"].values())/len(employees["Ali"].values())
sara_avg = sum(employees["Sara"].values()) / len(employees["Sara"])
ahmed_avg = sum(employees["Ahmed"].values()) / len(employees["Ahmed"])
averages = {"Ali": ali_avrg, "Sara": sara_avg, "Ahmed": ahmed_avg}
top=max(averages,key=averages.get)
highest_sql=max(employees,key=employees.get("SQL"))
highest_sql
```

```python
# Q40
# Calculate total inventory value, find the most valuable product based on price × stock,
# and update the stock after selling 2 laptops and 5 mice.
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}
inventory_value=sum(items["price"] * items["stock"] for items in inventory.values())
inventory_value
mostvaluable=max(inventory.values(),key=lambda items:items["price"] * items["stock"] )
mostvaluable
inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5
```
