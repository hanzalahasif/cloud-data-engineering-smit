## Q10 — List Transformation Challenge

```python
data = [15, 5, 25, 10, 35, 20, 30]

# Build a list of the three largest unique values ascending,
# then the two smallest unique values descending.
# Expected: [25, 30, 35, 10, 5]
# Write your solution below.
three_largest_unique=[]
first=max(set(data))
three_largest_unique.insert(0,first)
data.remove(first)
second=max(set(data))
three_largest_unique.insert(1,second)
data.remove(second)
third=max(set(data))
three_largest_unique.insert(2,third)
three_largest_unique.sort()

three_largest_unique

```

```python
three_smallest_unique = []
first = min(set(data))
three_smallest_unique.insert(0, first)
data.remove(first)
second = min(set(data))
three_smallest_unique.insert(1, second)
three_smallest_unique.sort()
print(three_smallest_unique)
final_list=three_largest_unique+three_smallest_unique
```
